```python
import json
import pandas as pd
```

```python
FILE_PATH = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\YourLibrary.json"

# Leer el archivo indicando codificación UTF-8
with open(FILE_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

# Convertir la lista "tracks" a DataFrame
Library = pd.DataFrame(data["tracks"])

# Mostrar los primeros 5 registros
print(Library.head())
```

```python
# Ranking de artistas en mi librería de Spotify según la cantidad de canciones que tengo de cada uno
ranking_artistas = (
    Library.groupby("artist")
    .size()
    .reset_index(name="cantidad_canciones")
    .sort_values(by="cantidad_canciones", ascending=False)
)

ranking_artistas
```

```python
# Rutas de los archivos
FILE_PATH_2 = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\StreamingHistory_music_0.json"
FILE_PATH_3 = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\StreamingHistory_music_1.json"

with open(FILE_PATH_2, "r", encoding="utf-8") as f:
    data_0 = json.load(f)

with open(FILE_PATH_3, "r", encoding="utf-8") as f:
    data_1 = json.load(f)

# Cargar en DataFrames y concatenar
df0 = pd.DataFrame(data_0)
df1 = pd.DataFrame(data_1)

streaming_history = pd.concat([df0, df1], ignore_index=True)

# Convertir milisegundos a minutos para facilitar el análisis
streaming_history["minutesPlayed"] = (
    streaming_history["msPlayed"] / 60000
).round(2)

# Mostrar las primeras filas de la tabla combinada
streaming_history.head()
```

```python
# Contar reproducciones para TODOS los artistas ordenados de mayor a menor
ranking_completo = (
    streaming_history["artistName"].value_counts().reset_index()
)
ranking_completo.columns = ["Artista", "Cantidad_de_Reproducciones"]

# Ver el DataFrame completo
ranking_completo
```
