```python
import json
import pandas as pd
```

```python
FILE_PATH = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\YourLibrary.json"

# Leer el archivo indicando codificación UTF-8
with open(FILE_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

# Convertir la lista "tracks" a DataFrame
Library = pd.DataFrame(data["tracks"])

# Mostrar los primeros 5 registros
print(Library.head())
```

```python
# Ranking de artistas en mi librería de Spotify según la cantidad de canciones que tengo de cada uno
ranking_artistas = (
    Library.groupby("artist")
    .size()
    .reset_index(name="cantidad_canciones")
    .sort_values(by="cantidad_canciones", ascending=False)
)

ranking_artistas
```

```python
FILE_PATH_2 = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\StreamingHistory_music_0.json"
FILE_PATH_3 = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\StreamingHistory_music_1.json"

with open(FILE_PATH_2, "r", encoding="utf-8") as f:
    data_0 = json.load(f)

with open(FILE_PATH_3, "r", encoding="utf-8") as f:
    data_1 = json.load(f)

df0 = pd.DataFrame(data_0)
df1 = pd.DataFrame(data_1)
streaming_history = pd.concat([df0, df1], ignore_index=True)

# Convertir milisegundos a minutos
streaming_history["minutesPlayed"] = (
    streaming_history["msPlayed"] / 60000
).round(2)

# Crear la tabla del ranking agrupado por artista
ranking_completo_detalle = (
    streaming_history.groupby("artistName")
    .agg(
        cantidad_reproducciones=("trackName", "count"),
        minutos_totales=("minutesPlayed", "sum"),
    )
    .reset_index()
    .sort_values(by="cantidad_reproducciones", ascending=False)
)

# Redondear minutos
ranking_completo_detalle["minutos_totales"] = ranking_completo_detalle[
    "minutos_totales"
].round(2)

# Guardar en formato CSV
ranking_completo_detalle.to_csv(
    "Ranking_Artistas_Spotify.csv", index=False, encoding="utf-8-sig"
)

print("¡Archivo CSV guardado con éxito en tu computadora!")

#  Mostrar las primeras filas en pantalla
ranking_completo_detalle.head(10)
```

```python
#  Crear el ranking y filtrar los PRIMEROS 100
top_100_artistas = (
    streaming_history.groupby("artistName")
    .agg(
        cantidad_reproducciones=("trackName", "count"),
        minutos_totales=("minutesPlayed", "sum"),
    )
    .reset_index()
    .sort_values(by="cantidad_reproducciones", ascending=False)
    .head(100)  # 👈 Filtra solo el Top 100
)

# Redondear minutos a 2 decimales
top_100_artistas["minutos_totales"] = top_100_artistas["minutos_totales"].round(2)

# Guardar la tabla del Top 100 en un archivo CSV
ruta_salida = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\Top100_Artistas_Spotify.csv"

top_100_artistas.to_csv(ruta_salida, index=False, encoding="utf-8-sig")

print(f"¡Top 100 guardado exitosamente en:\n{ruta_salida}")
```

```python
import json
import pandas as pd

# 1. Cargar archivos
FILE_PATH_2 = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\StreamingHistory_music_0.json"
FILE_PATH_3 = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\StreamingHistory_music_1.json"

with open(FILE_PATH_2, "r", encoding="utf-8") as f:
    data_0 = json.load(f)

with open(FILE_PATH_3, "r", encoding="utf-8") as f:
    data_1 = json.load(f)

streaming_history = pd.concat(
    [pd.DataFrame(data_0), pd.DataFrame(data_1)], ignore_index=True
)

# 2. Convertir endTime a formato datetime
streaming_history["endTime"] = pd.to_datetime(streaming_history["endTime"])

# 3. Extraer columnas temporales
streaming_history["fecha"] = streaming_history["endTime"].dt.date
streaming_history["año_mes"] = streaming_history["endTime"].dt.to_period(
    "M"
)  # Ej: 2026-04
streaming_history["hora"] = streaming_history["endTime"].dt.hour
streaming_history["minutesPlayed"] = (
    streaming_history["msPlayed"] / 60000
).round(2)

# 4. Guardar archivo completo con fechas
ruta_salida_detalle = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\Streaming_Con_Fechas.csv"
streaming_history.to_csv(
    ruta_salida_detalle, index=False, encoding="utf-8-sig"
)

print(f"¡Archivo detallado guardado en:\n{ruta_salida_detalle}")
```
