```python
import json
import pandas as pd
```

```python
FILE_PATH = r"C:\Users\FernandaCorrea\Documents\Documentos Belén\Certificado avanzado Data Science - ITBA\Módulo 4 - Data Viz\my_spotify_data (2)\Spotify Account Data\YourLibrary.json"

# Leer el archivo indicando codificación UTF-8
with open(FILE_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

# Convertir la lista "tracks" a DataFrame
Library = pd.DataFrame(data["tracks"])

# Mostrar los primeros 5 registros
print(Library.head())
```

```python
# Ranking de artistas
ranking_artistas = (
    Library.groupby("artist")
    .size()
    .reset_index(name="cantidad_canciones")
    .sort_values(by="cantidad_canciones", ascending=False)
)

ranking_artistas
```
